### Which school districts have filed the most appeals?

```python
## create a new dataframe with only school district appeals included
sd_df = all_appeals[all_appeals['on_behalf_of']=='School District']
```

```python
## how many unique school districts are included?
sd_df.school_district_name.nunique()
```

```python
## how many appeals have school districts filed since tax year 2023?
sd_df.groupby('school_district_name')['_id'].count().sort_values(ascending=False).reset_index().head(20)
```

```python
## filed in 2026, per school district?
sd_df_26 = sd_df[sd_df['tax_year']==2026]
sd_df_26.groupby('school_district_name')['_id'].count().sort_values(ascending=False).reset_index().head(20)
```

```python
## filed in 2025, per school district?
sd_df_25 = sd_df[sd_df['tax_year']==2025]
sd_df_25.groupby('school_district_name')['_id'].count().sort_values(ascending=False).reset_index().head(20)
```

```python
## filed in 2024, per school district?
sd_df_24 = sd_df[sd_df['tax_year']==2024]
sd_df_24.groupby('school_district_name')['_id'].count().sort_values(ascending=False).reset_index().head(20)
```

```python
## filed in 2023, per school district?
sd_df_23 = sd_df[sd_df['tax_year']==2023]
sd_df_23.groupby('school_district_name')['_id'].count().sort_values(ascending=False).reset_index().head(20)
```

### Change over time

```python
## save as groups
sd_25_grouped = sd_df_25.groupby('school_district_name')['_id'].count().sort_values(ascending=False).reset_index()
sd_26_grouped = sd_df_26.groupby('school_district_name')['_id'].count().sort_values(ascending=False).reset_index()
sd_24_grouped = sd_df_24.groupby('school_district_name')['_id'].count().sort_values(ascending=False).reset_index()
sd_23_grouped = sd_df_23.groupby('school_district_name')['_id'].count().sort_values(ascending=False).reset_index()
```

```python
groups = {
    "sd_25_grouped": sd_25_grouped,
    "sd_23_grouped": sd_23_grouped,
    "sd_24_grouped": sd_24_grouped,
    "sd_26_grouped": sd_26_grouped,
}

for name, group in groups.items():
    group.rename(columns={"_id": name}, inplace=True)
```

```python
df_list = [sd_26_grouped,sd_25_grouped,sd_24_grouped,sd_23_grouped]
df_list_indexed = [df.set_index(df.columns[0]) for df in df_list]
```

```python
all_dfs = pd.concat(df_list_indexed, axis=1).fillna(0)
all_dfs.head(20)
```

```python
## find change between '25 and '26
all_dfs['chg_25_26'] = ((all_dfs['sd_26_grouped'] - all_dfs['sd_25_grouped'])/all_dfs['sd_25_grouped']).round(2)
```

```python
all_dfs.sort_values(by='chg_25_26',ascending=False).head(20)
```
