```python
## import libraries
import pandas as pd
```

```python
## import data
appeals_26 = pd.read_csv('input/90432617-9b09-4084-919b-02aa002b6512.csv')
appeals_25 = pd.read_csv('input/c74b5a54-7448-4aa2-82f9-e6663fc10412.csv')
appeals_24 = pd.read_csv('input/12e00874-bdca-46c2-89ab-bd0e9272b3cb.csv')
appeals_23 = pd.read_csv('input/1547be15-34a6-4994-b73d-20bcf4718dba.csv')
```

```python
## create one large df
df_list = [appeals_26,appeals_25, appeals_24,appeals_23]
all_appeals = pd.concat(df_list,ignore_index=True)
```

```python
## strip muni column of white space
all_appeals['municipality'] = all_appeals['municipality'].str.strip()
```

## Analysis

### How many appeals according to class?

```python
## how many appeals according to class?
all_appeals.groupby(['tax_year','class'])['parcel_id'].count().tail(20)
```

### According to muni?

```python
## according to muni?
all_appeals.groupby(['tax_year','municipality'])['parcel_id'].count().sort_values(ascending=False).head(20)
```

### According to complainant, muni?

```python
## according to complainant?
all_appeals.groupby(['tax_year','municipality','on_behalf_of'])['parcel_id'].count().sort_values(ascending=False).head(10)
```

### Mt. Lebanon only

```python
mt_leb = all_appeals[all_appeals['municipality']=='Mt.Lebanon']
```

```python
mt_leb.to_csv('output/mt_leb_full.csv')
```
